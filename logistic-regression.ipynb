```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression

# Read CSV
data = pd.read_csv("test_data.csv")

# Remove missing Age values
data = data.dropna(subset=["Age"])

# Input and output
X = data[["Age"]]
y = data["Survived"]

# Logistic Regression
model = LogisticRegression()
model.fit(X, y)

# Predictions
data["Prediction"] = model.predict(X)

# Plot actual data
plt.scatter(data["Age"], data["Survived"], label="Actual")

# Plot logistic regression curve
age_range = pd.DataFrame({
    "Age": range(int(data["Age"].min()), int(data["Age"].max()) + 1)
})

probability = model.predict_proba(age_range)[:, 1]

plt.plot(age_range["Age"], probability, label="Logistic Regression")

plt.xlabel("Age")
plt.ylabel("Survived (0 = Died, 1 = Survived)")
plt.title("Age vs Survival")
plt.legend()
plt.show()
```
