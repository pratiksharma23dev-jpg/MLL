```python
import pandas as pd
import matplotlib.pyplot as plt

# Read data
data = pd.read_csv("heightweight.csv")

# Take Height and Weight
x = data["Height"]
y = data["Weight"]

# Calculate means
x_mean = sum(x) / len(x)
y_mean = sum(y) / len(y)

# Calculate slope (b1)
numerator = 0
denominator = 0

for i in range(len(x)):
    numerator += (x[i] - x_mean) * (y[i] - y_mean)
    denominator += (x[i] - x_mean) ** 2

b1 = numerator / denominator

# Calculate intercept (b0)
b0 = y_mean - b1 * x_mean

print("Slope:", b1)
print("Intercept:", b0)

# Predict weight
height = 175
predicted_weight = b0 + b1 * height

print("Predicted weight:", predicted_weight)

# Plot
predicted = []

for i in range(len(x)):
    predicted.append(b0 + b1 * x[i])

plt.scatter(x, y)
plt.plot(x, predicted)
plt.xlabel("Height")
plt.ylabel("Weight")
plt.title("Linear Regression")
plt.show()
```
